## Convert TotalCharges to numeric

```python
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)
```

```python
print(df_clean["TotalCharges"].dtype)
print("Missing TotalCharges:", df_clean["TotalCharges"].isnull().sum())
```

## Task 5 — Check duplicates again

## Task 6 — Check for invalid numerical values

```python
print("Invalid tenure:", (df_clean["tenure"] < 0).sum())
print("Invalid MonthlyCharges:", (df_clean["MonthlyCharges"] < 0).sum())
print("Invalid TotalCharges:", (df_clean["TotalCharges"] < 0).sum())
```

## Task 7 — Check for inconsistent text formatting

```python
for col in df_clean.select_dtypes(include="object").columns:
    if df_clean[col].str.contains(r"^\s|\s$", regex=True, na=False).any():
        print("Extra spaces found in:", col)
```
